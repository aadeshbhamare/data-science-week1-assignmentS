```python

# ==========================================
# Fictional E-commerce Sales Data Analysis
# ==========================================
# 1. Data Manipulation and Numerical Operations
import os
import shutil
import pandas as pd
import numpy as np
# 2. Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns
# 3. Date and Time Handling
from datetime import datetime
# 4. File and Directory Handling
import os
# 5. Warnings Management
import warnings
warnings.filterwarnings("ignore")
# 6. Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 1000)
# Set visualization style
sns.set_theme(style="whitegrid")
print("All libraries imported successfully!")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Setup completed successfully!")
```

```python
missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(missing_report)
```

```python
# STEP 7: DUPLICATE ANALYSIS
print("Exact Duplicate Rows:", df.duplicated().sum())
order_cols = [
    col for col in df.columns
    if "order" in col.lower() and "id" in col.lower()
]
if order_cols:
    order_col = order_cols[0]
    print("Possible Order ID Column:", order_col)
    print("Repeated Order ID Entries:",
          df[order_col].duplicated().sum())
else:
    print("No obvious Order ID column found.")
```

```python
# STEP 5: NUMERICAL DATA VALIDATION
print("Negative Quantities:",
      (df["Quantity"] < 0).sum())
print("Zero Quantities:",
      (df["Quantity"] == 0).sum())
print("Negative Unit Prices:",
      (df["Unit Price"] < 0).sum())
print("Zero Unit Prices:",
      (df["Unit Price"] == 0).sum())
print("Negative Total Prices:",
      (df["Total Price"] < 0).sum())
print("Zero Total Prices:",
      (df["Total Price"] == 0).sum())
```

```python
# STEP 6: ORDER DATE VALIDATION
print("Original Date Data Type:",
      df["Order Date"].dtype)
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)
print("\nConverted Date Data Type:",
      df["Order Date"].dtype)
print("Invalid Dates:",
      df["Order Date"].isna().sum())
print("Earliest Order Date:",
      df["Order Date"].min())
print("Latest Order Date:",
      df["Order Date"].max())
```

```python
# STEP 7: CATEGORY AND STATUS VALIDATION
print("Product Category Counts:")
display(df["Product Category"].value_counts())
print("\nPayment Type Counts:")
display(df["Payment Type"].value_counts())
print("\nOrder Status Counts:")
display(df["Order Status"].value_counts())
print("\nExtra Spaces Check:")
for col in ["Product Category", "Payment Type", "Order Status"]:
    extra_spaces = (
        df[col].astype(str) != df[col].astype(str).str.strip()
    ).sum()
    print(f"{col}: {extra_spaces} values with extra spaces")
```

```python
# STEP 8: TOTAL PRICE VALIDATION

df["Expected Total"] = (
    df["Quantity"] * df["Unit Price"]
)
df["Price Difference"] = (
    df["Total Price"] - df["Expected Total"]
).abs()
tolerance = 0.01
mismatched_rows = (
    df["Price Difference"] > tolerance
).sum()
print("Total Rows:", len(df))
print("Price Mismatches:", mismatched_rows)
print("\nSample Price Comparison:")
display(
    df[
        ["Quantity", "Unit Price",
         "Total Price", "Expected Total"]
    ].head(10)
)
```

```python
# STEP 9: CREATE CLEANED DATASET
df_clean = df.copy()
# Remove temporary validation columns
df_clean = df_clean.drop(
    columns=["Expected Total", "Price Difference"],
    errors="ignore"
)
# Standardize column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.lower()
)
print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", df_clean.shape)
print("\nCleaned Column Names:")
print(df_clean.columns.tolist())
print("\nFirst 5 Cleaned Records:")
display(df_clean.head())
```

```python
# STEP 10: SAVE CLEANED DATASET

output_path = "/content/cleaned_sales_dataset.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("File path:", output_path)
print("Total rows:", len(df_clean))
print("Total columns:", len(df_clean.columns))
```

```python
# STEP 14: DATA CLEANING SUMMARY

cleaning_summary = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Exact Duplicate Rows",
        "Invalid Dates",
        "Negative Quantities",
        "Zero Unit Prices",
        "Price Mismatches"
    ],
    "Result": [
        len(df_clean),
        len(df_clean.columns),
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum(),
        df_clean["order_date"].isnull().sum(),
        (df_clean["quantity"] < 0).sum(),
        (df_clean["unit_price"] == 0).sum(),
        (
            (
                df_clean["total_price"]
                - df_clean["quantity"] * df_clean["unit_price"]
            ).abs() > 0.01
        ).sum()
    ]
})

display(cleaning_summary)
```

```python
# STEP 15: TOTAL REVENUE

total_revenue = df_clean["total_price"].sum()

print("Total Revenue:", round(total_revenue, 2))
```

```python
# STEP 16: TOTAL ORDERS

total_orders = df_clean["order_id"].nunique()

print("Total Orders:", total_orders)
```

```python
# STEP 17: AVERAGE ORDER VALUE (AOV)

total_revenue = df_clean["total_price"].sum()
total_orders = df_clean["order_id"].nunique()

aov = total_revenue / total_orders

print("Total Revenue:", round(total_revenue, 2))
print("Total Orders:", total_orders)
print("Average Order Value (AOV):", round(aov, 2))
```

```python
# STEP 18: TOTAL UNITS SOLD

total_units_sold = df_clean["quantity"].sum()

print("Total Units Sold:", total_units_sold)
```

```python
# STEP 19: TOP-SELLING PRODUCT CATEGORY

category_sales = (
    df_clean.groupby("product_category")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("Units Sold by Product Category:")
display(category_sales)

print("Top-Selling Category:", category_sales.idxmax())
print("Units Sold:", category_sales.max())
```

```python
# STEP 20: TOP REVENUE-GENERATING PRODUCT

product_revenue = (
    df_clean.groupby("product_id")["total_price"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Products by Revenue:")
display(product_revenue.head(10))

print("\nTop Revenue-Generating Product:", product_revenue.idxmax())
print("Total Revenue Generated:", round(product_revenue.max(), 2))
```

```python
# STEP 21: MONTHLY SALES TREND

df_clean["month"] = df_clean["order_date"].dt.to_period("M")

monthly_sales = (
    df_clean.groupby("month")["total_price"]
    .sum()
)

print("Monthly Sales:")
display(monthly_sales)
```

```python
# STEP 22: MONTHLY SALES LINE CHART

monthly_sales.index = monthly_sales.index.astype(str)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend - 2023")
plt.xlabel("Month")
plt.ylabel("Total Sales Revenue")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# STEP 23: CATEGORY-WISE REVENUE ANALYSIS

category_revenue = (
    df_clean.groupby("product_category")["total_price"]
    .sum()
    .sort_values(ascending=False)
)

# Display full numbers instead of scientific notation
pd.set_option("display.float_format", "{:,.2f}".format)

print("Revenue by Product Category:")
display(category_revenue.to_frame())

print("\nTop Revenue Category:", category_revenue.idxmax())
print("Total Revenue:", f"{category_revenue.max():,.2f}")
```

```python
# STEP 24: CATEGORY-WISE REVENUE BAR CHART

plt.figure(figsize=(10, 6))

category_revenue.plot(kind="bar")

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45, ha="right")
plt.ticklabel_format(axis="y", style="plain")
plt.tight_layout()
plt.show()
```

```python
# STEP 25: ORDER STATUS ANALYSIS

status_counts = df_clean["order_status"].value_counts()

print("Order Status Counts:")
display(status_counts.to_frame(name="Number of Orders"))

print("\nMost Common Order Status:", status_counts.idxmax())
print("Number of Orders:", status_counts.max())
```

```python
# STEP 26: ORDER STATUS PIE CHART

status_counts = df_clean["order_status"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Order Status Distribution")
plt.tight_layout()
plt.show()
```

```python
# STEP 27: PAYMENT TYPE ANALYSIS

payment_counts = df_clean["payment_type"].value_counts()

print("Orders by Payment Type:")
display(payment_counts.to_frame(name="Number of Orders"))

print("\nMost Used Payment Type:", payment_counts.idxmax())
print("Number of Orders:", payment_counts.max())
```

```python
# STEP 28: PAYMENT TYPE BAR CHART

payment_counts = df_clean["payment_type"].value_counts()

plt.figure(figsize=(9, 5))

payment_counts.plot(kind="bar")

plt.title("Orders by Payment Type")
plt.xlabel("Payment Type")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# STEP 29: CUSTOMER INSIGHTS

customer_analysis = (
    df_clean.groupby("customer_id")
    .agg(
        total_orders=("order_id", "nunique"),
        total_spent=("total_price", "sum"),
        total_units=("quantity", "sum")
    )
    .sort_values("total_spent", ascending=False)
)

print("Top 10 Customers by Total Spending:")
display(customer_analysis.head(10))

print("\nTotal Unique Customers:", df_clean["customer_id"].nunique())
print("Highest Customer Spending:", f"{customer_analysis['total_spent'].max():,.2f}")
```

```python
# STEP 30: WEEKLY SALES TREND

weekly_sales = (
    df_clean
    .set_index("order_date")["total_price"]
    .resample("W")
    .sum()
)

print("Weekly Sales Trend:")
display(weekly_sales.to_frame(name="Total Sales"))
```

```python
# STEP 31: WEEKLY SALES LINE CHART

plt.figure(figsize=(12, 5))

plt.plot(
    weekly_sales.index,
    weekly_sales.values,
    marker="o"
)

plt.title("Weekly Sales Trend - 2023")
plt.xlabel("Week")
plt.ylabel("Total Sales Revenue")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# STEP 32: BUSINESS INSIGHTS

# 1. Highest revenue category
top_category = category_revenue.idxmax()
top_category_revenue = category_revenue.max()

# 2. Lowest revenue category
low_category = category_revenue.idxmin()
low_category_revenue = category_revenue.min()

# 3. Highest sales month
top_month = monthly_sales.idxmax()
top_month_sales = monthly_sales.max()

# 4. Lowest sales month
low_month = monthly_sales.idxmin()
low_month_sales = monthly_sales.min()

# 5. Most common order status
top_status = status_counts.idxmax()
top_status_count = status_counts.max()

# 6. Most used payment type
top_payment = payment_counts.idxmax()
top_payment_count = payment_counts.max()

# 7. Highest spending customer
top_customer = customer_analysis["total_spent"].idxmax()
top_customer_spending = customer_analysis["total_spent"].max()

# Display insights
print("BUSINESS INSIGHTS")
print("-" * 50)

print(f"1. Highest revenue category: {top_category} "
      f"({top_category_revenue:,.2f})")

print(f"2. Lowest revenue category: {low_category} "
      f"({low_category_revenue:,.2f})")

print(f"3. Highest sales month: {top_month} "
      f"({top_month_sales:,.2f})")

print(f"4. Lowest sales month: {low_month} "
      f"({low_month_sales:,.2f})")

print(f"5. Most common order status: {top_status} "
      f"({top_status_count} orders)")

print(f"6. Most used payment type: {top_payment} "
      f"({top_payment_count} orders)")

print(f"7. Highest recorded customer spending: {top_customer} "
      f"({top_customer_spending:,.2f})")
```
